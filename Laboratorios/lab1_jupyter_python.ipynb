## Parte 1: Cómo usar Jupyter

Esta página se llama un **Jupyter Notebook**. Es un lugar para escribir código, ver resultados, y también escribir texto explicativo.

### Celdas de texto

En un notebook, cada rectángulo que contiene texto o código se llama una *celda*.

Las celdas de texto (como esta) se pueden editar haciendo doble clic. Están escritas en formato [Markdown](https://www.markdownguide.org/basic-syntax/), que permite agregar formato y encabezados. No necesitas aprender Markdown, pero puede ser útil.

Después de editar una celda de texto, haz clic en el botón "ejecutar celda" (▶) o presiona `Shift` + `Return` para confirmar los cambios.

### Celdas de código

Las otras celdas contienen código en Python 3. Ejecutar una celda de código correrá todo el código que contiene.

Para ejecutar el código en una celda, primero haz clic en esa celda para activarla. Luego presiona ▶ o `Shift` + `Return`.

Prueba ejecutando esta celda:

```python
print("¡Hola, Mundo!")
```

Cuando ejecutas varias celdas, el orden importa. Observa el orden del output al ejecutar la siguiente celda:

```python
print("Primero se imprime esta línea,")
print("y luego esta.")
```

## Parte 2: NumPy básico

`NumPy` es el módulo de cómputo numérico de Python. Es la base de Pandas y de casi todas las herramientas de ciencia de datos que usaremos en el curso.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print('✅ Librerías importadas correctamente')
```

### Arrays

El elemento central de NumPy es el **array**. Similar a una lista de Python, pero más eficiente para operaciones numéricas.

Usamos `np.array()` para crear uno:

```python
# Crear un array
arr = np.array([1, 2, 3, 4, 5])
arr
```

Podemos acceder a sus atributos:

```python
print('Forma:', arr.shape)
print('Tipo de datos:', arr.dtype)
print('Elemento en posición 2:', arr[2])
print('Elementos del 1 al 3:', arr[1:4])
```

### Operaciones vectorizadas

Una de las ventajas de NumPy es que las operaciones se aplican elemento a elemento automáticamente:

```python
# Con listas Python, + concatena
[1, 2, 3] + [4, 5, 6]
```

```python
# Con arrays NumPy, + suma elemento a elemento
np.array([1, 2, 3]) + np.array([4, 5, 6])
```

### Velocidad: listas vs arrays

NumPy es mucho más rápido que las listas de Python para operaciones numéricas grandes. Veamos la diferencia:

```python
muestra_lista = list(range(1_000_000))
muestra_array = np.arange(1_000_000)
```

```python
%%time
# Con lista Python
[x**2 for x in muestra_lista]
;
```

```python
%%time
# Con array NumPy
muestra_array**2
;
```

**¿Qué observas?** NumPy puede ser 10-50x más rápido. Cuando trabajemos con millones de registros del IMSS o del Banco Mundial, esta diferencia importa.

## Parte 3: Primera visualización

Matplotlib es la librería base de visualización en Python. Aquí haremos una gráfica simple con datos económicos reales.

### Ejercicio 2: Graficar inflación en México

Usando los datos de inflación anual de México de 2013 a 2022, crea una gráfica de línea.

```python
# Datos de inflación anual en México (INEGI)
años = np.array([2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022])
inflacion = np.array([3.97, 4.02, 2.13, 3.36, 6.77, 4.83, 2.83, 3.40, 7.36, 7.82])

# Graficar
plt.figure(figsize=(10, 5))
plt.plot(años, inflacion, marker='o', color='steelblue', linewidth=2)

# Línea de referencia en 4% (meta del Banxico)
plt.axhline(y=4, color='red', linestyle='--', alpha=0.7, label='Meta Banxico (4%)')

plt.title('Inflación anual en México 2013–2022')
plt.xlabel('Año')
plt.ylabel('Inflación (%)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
