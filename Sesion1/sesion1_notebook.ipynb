## Parte 2 — Tipos de Datos

### Datos estructurados vs no estructurados

| Tipo | Descripción | Ejemplos en economía |
|------|-------------|----------------------|
| **Estructurados** | Organizados en filas y columnas | ENOE, ENIGH, registros IMSS, datos panel |
| **No estructurados** | Sin formato tabular predefinido | Contratos, discursos de bancos centrales, imágenes satelitales |

En este curso trabajamos principalmente con datos **estructurados**.

### Tipos de variables

Dentro de una tabla, cada columna puede ser de distinto tipo:

| Tipo | Descripción | Ejemplo en economía |
|------|-------------|---------------------|
| **Numérica continua** | Cualquier valor en un rango | PIB, salario, inflación |
| **Numérica discreta** | Solo valores enteros | Número de hijos, años de escolaridad |
| **Categórica nominal** | Categorías sin orden | País, sector económico, sexo |
| **Categórica ordinal** | Categorías con orden | Nivel de ingreso (bajo/medio/alto) |
| **Booleana** | Verdadero / Falso | ¿Tiene empleo formal? ¿Recibe subsidio? |
| **Fecha/Tiempo** | Marca temporal | Trimestre, año fiscal, fecha de registro |

### Formatos de almacenamiento

Los datos estructurados pueden vivir en distintos formatos. Los más comunes:

| Formato | Cuándo usarlo |
|---------|---------------|
| **CSV** | Datos tabulares pequeños o medianos, fácil de compartir |
| **TSV** | Igual que CSV pero separado por tabuladores |
| **JSON** | Datos de APIs, estructuras anidadas |
| **SQL** | Datos relacionales, consultas complejas |
| **Parquet** | Big data, lectura eficiente de columnas |

### Herramientas según escala

| Escala | Herramienta | Ejemplo |
|--------|-------------|--------|
| < 100k filas | CSV + Pandas | Mayoría de análisis de este curso |
| 100k – 10M filas | SQL / Parquet + Pandas | Microdatos ENOE completos |
| > 10M filas | Spark / BigQuery | Registros SAT nacionales |
| Datos en línea | APIs + JSON | Banxico, Banco Mundial |

Veamos estos formatos en acción con el mismo dataset en CSV, TSV y JSON:

```python
# El mismo dataset en tres formatos distintos
base_url = 'https://raw.githubusercontent.com/DS-100/course-notes/main/content/eda/data/'

# CSV — Comma Separated Values
elections_csv = pd.read_csv(base_url + 'elections.csv')
print(f'CSV: {elections_csv.shape[0]} filas, {elections_csv.shape[1]} columnas')
elections_csv.head(3)
```

```python
# TSV — Tab Separated Values (mismo dato, distinto delimitador)
elections_tsv = pd.read_csv(base_url + 'elections.txt', sep='\t')
print(f'TSV: {elections_tsv.shape[0]} filas, {elections_tsv.shape[1]} columnas')
elections_tsv.head(3)
```

```python
# JSON — JavaScript Object Notation
elections_json = pd.read_json(base_url + 'elections.json')
print(f'JSON: {elections_json.shape[0]} filas, {elections_json.shape[1]} columnas')
elections_json.head(3)
```

## Parte 3 — Una segunda base: indicadores del Banco Mundial

Ahora cargamos una base más rica con múltiples indicadores por país.
Esta es la misma base que usaremos en sesiones futuras de visualización.

```python
# Metadata de países (incluye región, nivel de ingreso, etc.)
countries = wb.get_countries()

# Indicadores para todos los países, último año disponible
indicadores = wb.download(
    indicator={
        'NY.GDP.PCAP.CD': 'gdp_per_capita',
        'SL.UEM.TOTL.ZS': 'unemployment',
        'SP.DYN.LE00.IN': 'life_expectancy',
        'SP.POP.TOTL':    'population',
        'FP.CPI.TOTL.ZG': 'inflation',
        'SI.POV.GINI':    'gini'
    },
    country='all',
    start=2020,
    end=2022
).reset_index()

# Cruzamos con metadata para tener región y nivel de ingreso
df = indicadores.merge(
    countries[['name','region']],
    left_on='country',
    right_on='name'
)

print(f"Base cargada: {df.country.nunique()} paises y 6 indicadores")
```

```python
# Veamos qué hay
df.head()
```

```python
# ¿Qué columnas (indicadores) tiene?
print("Columnas disponibles:")
for col in df.columns:
    print(f"  - {col}")
```

```python
# Estadísticas descriptivas básicas
# ¿Qué observan? ¿Hay valores raros?
df.describe()
```

```python
# ¿Hay datos faltantes?
print("Valores faltantes por columna:")
print(df.isnull().sum())
```
